# Bayesian Inference: Analysis Overview

## 1. Position in the paper and the question being asked

This section sits at the end of the paper, after the PINT pulsar-timing analysis of Section 4 has produced the principal measurement of this work: $\dot P_{\rm obs}=0\pm 1.7\times 10^{-15}$ s/s. That precision is roughly $1000\times$ tighter than the periodogram-based upper limit from Salcedo et al. (2024). The methodological argument of the paper — that non-periodogram, phase-coherent timing techniques deliver dramatic precision gains on fast-spinning CVs — is established by the time we get here.

What this section asks is: *given that precision, what physical insight does the new $\dot P$ buy us?* The original hypothesis, motivated by the explicit $\mu^2/r_m^3$ scaling of the IP magnetic-torque term in Bhattacharyya & Chakrabarty 2017, was that the new $\dot P$ would tighten the magnetic-field posterior beyond what spectral data alone can constrain. We test this hypothesis with a unified Bayesian inference combining timing ($\dot P$) and spectral ($L_X$, $kT_{\rm max}$) observables under broad population priors, and we follow it with a validation notebook designed to expose the result to the full set of independent cross-checks that a 4-parameter / 3-observable inference deserves.

The headline result of the section is that the hypothesis is **empirically false**: re-running the inference with $\dot P$ excluded produces posteriors indistinguishable from the with-$\dot P$ run. We do not present this as a failure. It is the section's principal scientific finding, and the validation notebook turns it into a constructive statement: the prior-predictive distribution of $|\dot P|$ is naturally narrow because $\dot P\propto P^2/I$ and $P=29.6$ s makes the natural scale tiny, so most spectrally-allowed parameter combinations already produce $|\dot P|\le \sigma_{\dot P}$. There is simply nothing left for the timing likelihood to down-weight. This explains *why* a high-precision $\dot P$ on this particular system fails to add parameter leverage, and turns the result into a generalisable lesson for the broader CV-timing community: precision timing pays off in propeller spin-down or strong spin-up systems where the predicted $|\dot P|$ exceeds the achievable measurement floor, and degrades to a consistency check (rather than a constraint) for equilibrium accretors like CTCV J2056.

The section is therefore both the answer to the paper's "so what?" question and an honest statement about when the methods of Sections 3-4 should be expected to translate into physical parameter improvements vs. when they should be marketed as falsifiability tests.

## 2. Model choice — IP only, with no propeller and no nmCV competitor

CJ2056 is modelled as an **accreting intermediate polar** throughout this section. Three modelling choices deserve explicit motivation, since each ruled out an alternative that earlier iterations of this analysis (notebook `07`) entertained.

**IP only, no nmCV competitor.** Notebook `07` ran a Bayes-factor model comparison between IP and a non-magnetic CV (boundary-layer accretion to $R_{\rm WD}$). The current analysis dispenses with the nmCV branch and runs an IP-only inference. The classification is no longer in question: Salcedo et al. (2024, §6.1-6.2) report direct evidence for a magnetically channeled accretion column from hard X-ray spectra and from phase-resolved hardness-ratio modulation; our own NICER 2024 hardness-ratio analysis (Section 3.2) finds the modulation at $6.7\sigma$ significance, anti-correlated with flux as expected for an accretion-curtain absorber (Rosen et al. 1988). Computing a Bayes factor in this section would still produce a number, but it would be a number on a question the data have already answered. The paper's argument is also not about classification; it is about parameter inference under the IP model. Dropping nmCV simplifies the prior space, eliminates an Occam-penalty asymmetry that depends on the somewhat-arbitrary 6-decade $\log_{10}\mu$ prior, and lets us put all of the methodological emphasis on the $\dot P$-leverage question.

**No propeller.** The IP forward models reject the propeller regime ($r_m > r_{\rm co}$) at the likelihood level by returning NaN, which the sampler maps to $\ln\mathcal L = -10^{100}$. This is a hard prior. Two physical arguments justify it. First, propeller systems eject infalling matter at $r_m$ and do not produce a standing accretion shock on the WD surface; the observed $kT\sim 16$ keV / $L_X\sim 10^{31}$ erg/s emission is unambiguously an accretion-shock signature. Second, the AE Aqr-class propeller prototypes have very different X-ray spectral and timing phenomenology. A softer treatment — say, a propeller branch with no $kT$/$L_X$ likelihood term — would degenerately admit propeller solutions because the $\dot P=0$ likelihood by itself cannot down-weight them; an earlier version of this analysis hit exactly that failure mode. Hard exclusion is the cleaner choice.

**No alternative torque prescriptions.** The Bhattacharyya & Chakrabarty 2017 formula is one of several disk-torque prescriptions in the literature (Ghosh & Lamb 1979, Wang 1995, Rappaport et al. 2004). All of them produce a torque curve that crosses zero at some $r_m/r_{\rm co}<1$, and the *qualitative* near-equilibrium statement ("net torque is small, system is near its equilibrium fastness") is robust across them. The quantitative numerical value of $r_m/r_{\rm co}$ at equilibrium and the inferred $\mu$ (and thus $B$) carry an additional $\mathcal O(10\text{-}30\%)$ systematic from this prescription choice that is not propagated; we flag it in §10.

## 3. Observational inputs

The inference ingests three observables, each treated as an independent Gaussian in the likelihood. The data are tabulated below; their derivation is referenced where it lives.

| Observable | Value | $1\sigma$ uncertainty | Source |
|:---|:---|:---|:---|
| $P$ (spin period) | $29.609685843$ s | $3.0\times 10^{-8}$ s | Section 4.5 (PINT WLS fit) |
| $\dot P_{\rm obs}$ | $0$ s/s | $1.7\times 10^{-15}$ s/s | Section 4.5 |
| $d$ (distance) | 261.6 pc | 7.4 pc | Gaia parallax (Salcedo 2024) |
| $F_X$ (unabs., 0.3-12 keV) | $2.2\times 10^{-12}$ erg s$^{-1}$ cm$^{-2}$ | 5% (statistical) | Salcedo 2024 |
| $L_X$ (derived) | $1.8\times 10^{31}$ erg s$^{-1}$ | $\sigma_{L_X}/L_X\approx 7.6\%$ | $4\pi d^2 F_X$, propagated |
| $kT_{\rm max}$ | 16.3 keV | 1.15 keV | Salcedo 2024 Table 6 (MKCFLOW) |

Three things deserve comment.

**$L_X$ uncertainty propagation.** $\sigma_{L_X}/L_X = \sqrt{(2\sigma_d/d)^2 + (\sigma_{F_X}/F_X)^2}$ — the distance enters squared in the area, hence the factor of 2. With $\sigma_d/d=2.8\%$ and $\sigma_{F_X}/F_X=5\%$, the dominant contribution is the flux uncertainty. Systematic errors from absorption, spectral-model choice, and instrument calibration could plausibly be 15-30%; we use the statistical-only number and absorb the difference into the $\dot M$ posterior implicitly.

**$kT_{\rm max}$ from MKCFLOW.** The reported 90% confidence interval is $16.3^{+1.9}_{-1.8}$ keV; we convert to a 1σ Gaussian via $\sigma\approx 1.9/1.645\approx 1.15$. MKCFLOW is the cooling-flow / boundary-layer model (intended for nmCVs), whereas Salcedo's MCVSPEC IP fits (Tab 7) do not directly tabulate $T_s$. We use the MKCFLOW $kT_{\rm max}$ in the IP forward model as a *model-independent* observable: $kT_{\rm max}$ is the high-energy continuum cutoff of optically thin bremsstrahlung emission, set by the maximum post-shock temperature, and is approximately invariant across the cooling-flow → magnetically-channeled-column geometry transition. A more rigorous treatment would refit Salcedo's spectra with a unified IP model; that is out of scope for this paper. The implication is a small additional model dependence in $kT_{\rm obs}$, not propagated.

**$\dot P_{\rm obs}=0$ exactly.** The PINT fit returned $\dot P=0\pm 1.7\times 10^{-15}$ s/s — i.e., consistent with zero, with $\dot P$ statistically insignificant under an $F$-vs-$F+\dot F$ comparison (Section 4.5). We use $\dot P_{\rm obs}=0$ in the likelihood; the effective $\dot P$ constraint is therefore "$|\dot P|\lesssim \sigma_{\dot P}$" rather than "$\dot P=$ a specific nonzero value". This matters for the leverage diagnostic (§8): the term cannot drive the posterior toward any particular non-trivial $\dot P$, only away from regions that produce $|\dot P|\gg \sigma_{\dot P}$.

**WD mass-radius-inertia relation.** A pre-tabulated $M$-$R(M)$-$I(M)$ table (zero-temperature C-12 WDs, range $\sim 0.01$-$1.4\,M_\odot$) is loaded once; we interpolate it cubically. The mass prior is clipped by $\pm 0.001\,M_\odot$ from the table edges to avoid cubic-spline edge effects. $R(M)$ enters every forward model and $I(M)$ enters $\dot P_{\rm pred}$.

## 4. Forward models

The accretion physics is the load-bearing piece of the inference; everything else is bookkeeping. All forward models share a single helper that returns $(M, \dot M, \mu, R, r_m, r_{\rm co})$ in CGS for a given parameter vector, including the standard Ghosh-Lamb magnetospheric and Keplerian-corotation radii:

$$r_m = \xi\!\left(\frac{\mu^4}{2 G M \dot M^2}\right)^{1/7}, \qquad r_{\rm co} = \left(\frac{G M P^2}{4\pi^2}\right)^{1/3}.$$

If the resulting $r_m$ falls outside $[R_{\rm WD}, r_{\rm co}]$ — i.e., either inside the WD surface (buried-field regime) or beyond corotation (propeller regime) — every forward model returns NaN and the likelihood rejects the sample.

**Spin derivative** (Bhattacharyya & Chakrabarty 2017, Eq. 9):

$$\dot P_{\rm pred} = -\frac{P^2}{2\pi I}\,(\tau_{\rm acc} + \tau_{\rm mag}),$$

$$\tau_{\rm acc} = \dot M\sqrt{G M\, r_m},\qquad \tau_{\rm mag} = \frac{\mu^2}{9\, r_m^3}\,(2x^3 - 6 x^{3/2} + 3),\qquad x \equiv r_m/r_{\rm co}.$$

$\tau_{\rm acc}$ is the accretion spin-up torque; $\tau_{\rm mag}$ is the net magnetic torque from threading field lines, which crosses zero at some $x_*<1$ (the "fastness" at which spin-up and spin-down balance). The $P^2/I$ prefactor is what makes $|\dot P|$ small for *any* parameter combination at a 30-second spinner: this is the equation-level explanation for the leverage finding in §8 and the observation in §9a.

**Shock temperature** (Rankine-Hugoniot strong shock, $\gamma=5/3$, free-fall from $r_m$ to $R_{\rm WD}$):

$$kT_{\rm pred} = \tfrac{3}{16}\,\mu_{\rm mol}\, m_H\, v_{\rm ff}^2,\qquad v_{\rm ff}^2 = 2 G M\!\left(\frac{1}{R_{\rm WD}} - \frac{1}{r_m}\right),$$

with $\mu_{\rm mol}=0.617$ for fully ionised solar composition. We do not apply the Yu et al. (2018) $\alpha_{\rm Yu}=0.646$ empirical correction here — that calibration was done on 11 nmCV boundary-layer shocks and does not transfer to the magnetically channeled column geometry of an IP. The unadjusted Rankine-Hugoniot result is the conservative choice.

**Band-corrected X-ray luminosity:**

$$L_{X,\rm pred} = f_{\rm band}(kT_{\rm pred})\, \cdot\, G M \dot M\!\left(\frac{1}{R_{\rm WD}} - \frac{1}{r_m}\right),\qquad f_{\rm band}(kT) = e^{-E_{\rm lo}/kT} - e^{-E_{\rm hi}/kT},$$

with $[E_{\rm lo}, E_{\rm hi}] = [0.3, 12]$ keV. $f_{\rm band}$ is the fraction of bolometric thermal-bremsstrahlung luminosity emitted in the $0.3$-$12$ keV $L_X$ band, assuming single-temperature emission with Gaunt factor $g_{\rm ff}\approx 1$. At $kT\sim 16$ keV, $f_{\rm band}\approx 0.5$ — the band correction is essential, not optional. Without it, the predicted $L_X$ overshoots the observed value by a factor $\sim 2$ at the best-fit point. The *implicit* assumption is single-temperature bremsstrahlung; the actual post-shock column is multi-temperature (hence Salcedo's MKCFLOW cooling-flow model), and this introduces a $\sim 20$-$30\%$ systematic on $L_{X,\rm pred}$ that the inference absorbs into the $\dot M$ posterior.

**Derived quantities** (from posterior samples, not sampled directly):

$$B = \mu / R_{\rm WD}^3,\qquad x = r_m / r_{\rm co},\qquad \tau_{\rm spin} = P / |\dot P_{\rm pred}|.$$

## 5. Priors

Priors are deliberately broad and population-based, with the design goal that the spectral observables enter only through the likelihood and not (via Salcedo's spectrally-derived posterior) through the prior.

| Parameter | Prior | Justification |
|:---|:---|:---|
| $M_{\rm WD}$ | $\mathcal N(0.81, 0.20)\,M_\odot$, clipped | Pala et al. (2022) volume-limited CV sample mean and dispersion |
| $\log_{10}\dot M$ | $\mathcal U(13, 18)$, log g/s | Spans dwarf-nova ($\sim 10^{13}$) to nova-like ($\sim 10^{17}$) |
| $\log_{10}\mu$ | $\mathcal U(29, 35)$, log G$\,$cm$^3$ | Brackets the full observed IP range |
| $\xi$ | log-$\mathcal U(0.3, 1.0)$ | Ghosh & Lamb (1979) disk-Alfvén range |

**Why broad pop priors instead of Salcedo's posterior.** A natural-seeming alternative would be to use Salcedo's joint $(M, \dot M)$ posterior as an informative prior. We do not, for two reasons. First, $kT_{\rm obs}$ enters our likelihood; if it had also gone into a $kT$-derived $M$ prior, we would be double-counting the same data. Second, Salcedo's $(M, \dot M)$ posterior carries an internal covariance from their joint spectral fit; collapsing it to a marginal prior would discard structure that our forward model can recover by combining $kT$ ($\propto GM/R$ at fixed $r_m/R$), $L_X$ ($\propto GM\dot M/R$ at fixed $r_m/R$), and the WD $M$-$R$ relation. The broad-prior + observable-likelihood approach reconstructs the right joint constraint through the physics.

**Why $\xi$ is sampled, not fixed.** Since $r_m\propto \xi$ and $\mu\propto r_m^{7/4}$ at fixed $(M, \dot M, r_m/R)$, $B$ scales as $\xi^{-7/4}$: a factor 2 in $\xi$ maps to a factor 3.4 in $B$. This is the dominant systematic on $B$. Fixing $\xi$ (as is conventional in MCVSPEC-style fits) reports a $B$ posterior that does not propagate the prefactor uncertainty; sampling $\xi$ on log-$\mathcal U(0.3, 1.0)$ marginalises it explicitly. The log-uniform choice (rather than uniform) is so that the prior is geometrically centred on the systematic rather than weighted toward the upper end.

**Mass clipping.** The $M_{\rm WD}$ prior is clipped to $[M_{\rm min,table}+0.001, M_{\rm max,table}-0.001]\,M_\odot$ to avoid cubic-spline edge artefacts in $R(M)$ and $I(M)$. The clip is far enough from the central $0.83\,M_\odot$ that it does not affect any reasonable posterior.

## 6. Likelihood and inference engine

All three observables enter as independent Gaussians:

$$\ln \mathcal L = -\tfrac12\!\left[\left(\frac{\dot P_{\rm obs} - \dot P_{\rm pred}}{\sigma_{\dot P}}\right)^2 + \left(\frac{kT_{\rm obs} - kT_{\rm pred}}{\sigma_{kT}}\right)^2 + \left(\frac{L_{X,\rm obs} - L_{X,\rm pred}}{\sigma_{L_X}}\right)^2\right].$$

Forward-model NaNs (i.e. $r_m\notin [R_{\rm WD}, r_{\rm co}]$) yield $\ln\mathcal L = -10^{100}$; non-positive predicted observables are treated the same way.

**Independence.** $kT$ and $L_X$ come from the same MKCFLOW spectral fit and share correlated systematics (most notably $N_H$). We do not propagate their covariance, which likely makes the joint constraint slightly tighter than it should be in the $kT$-$L_X$ direction; the effect is small relative to the $\xi$ prefactor systematic.

**Sampler.** dynesty `DynamicNestedSampler` with 2000 live points, multi-ellipsoidal bounding, rwalk proposal, fixed PCG64 seed for reproducibility. We use dynesty as a *posterior sampler*, not as an evidence calculator. Its motivating features here are clean handling of the hard $r_m\le r_{\rm co}$ likelihood boundary (NaN-rejection inside the likelihood is well-defined for nested sampling, where the algorithm only needs to compare likelihoods), automatic convergence diagnostics, and no burn-in tuning. We do not report Bayes factors in this section, so the choice of nested sampling over MCMC is for the boundary geometry rather than for $\ln Z$.

**Degrees of freedom.** With 4 sampled parameters and 3 observables, the system is formally under-constrained by data alone. The remaining degrees of freedom are absorbed by the population mass prior and the inequality constraint on $r_m$. The resulting $B$ posterior width (factor $\sim 4$ at $1\sigma$) is therefore an honest representation of limited information rather than a sampler artefact.

## 7. Main posterior results

After 1-3 minutes of sampling at $n_{\rm live}=2000$, the IP posterior is:

| Parameter | Median | 1σ interval |
|:---|:---|:---|
| $M_{\rm WD}$ | $0.87\,M_\odot$ | $+0.13\,/\,-0.08$ |
| $\log_{10}\dot M$ (g/s) | $14.72$ | $+0.03\,/\,-0.04$ |
| $\log_{10}\mu$ (G$\,$cm$^3$) | $30.09$ | $\pm 0.34$ |
| $\xi$ | $0.57$ | $+0.27\,/\,-0.19$ |
| $\log_{10}(B/\text{G})$ (derived) | $3.76$ | $\pm 0.30$ |
| $r_m/r_{\rm co}$ (derived) | $0.71$ | $+0.19\,/\,-0.20$ |
| $\tau_{\rm spin}=P/|\dot P|$ (derived) | $5.1$ Gyr | 1.8-10 Gyr |

The posterior tells a consistent physical story.

**Mass and accretion rate are tightly constrained.** $\dot M$ is pinned at the few-percent level by $L_X$ at fixed $(M, r_m/R)$; $M$ inherits a moderate constraint from the $kT(M, r_m/R)$ relation combined with the population prior. Both overlap Salcedo's published ranges (§9b).

**The magnetic moment, and hence $B$, is the loose parameter.** The $\xi$ marginal stays close to its log-uniform prior ($\xi=0.57^{+0.27}_{-0.19}$ vs prior log-mean $\approx 0.55$), and $\log_{10}\mu$ has a $\pm 0.34$ dex 1σ width. Together these give $\log_{10}(B/\text{G}) = 3.76\pm 0.30$, i.e. $B\sim 5.7$ kG with a factor-2 1σ envelope. This loose-but-honest $B$ posterior is the section's quantitative result on the magnetic field.

**Spin equilibrium.** $r_m/r_{\rm co}$ peaks near 0.7 and spans roughly 0.5-0.9 at 1σ. The right edge of the posterior is the propeller-prior boundary, *not* a peak: the posterior is being driven to the *interior* zero of the net torque rather than piling up at the cutoff. This is significant because in the Bhattacharyya & Chakrabarty 2017 prescription the magnetic torque crosses zero at some fastness $x_*<1$ — typically $x_*\sim 0.7$-$0.95$ depending on coupling details — so a posterior centred at $\sim 0.7$ is what one expects when the data are actually driving the system to its equilibrium fastness. The conclusion is best phrased as *consistent with near spin equilibrium* rather than *at* spin equilibrium; the data do not distinguish exactly-equilibrium from somewhat-sub-equilibrium accretion.

**Spin-evolution timescale.** $\tau_{\rm spin}\sim 5$ Gyr (1.8-10 Gyr at 68%) exceeds the Hubble time, consistent with long-term spin equilibrium expected for an evolved disk-fed IP.

## 8. The $\dot P$ leverage test (the section's headline finding)

The hypothesis that motivated this whole exercise is: *will the new $\dot P$ tighten the $B$ posterior beyond what spectral data alone constrain?* We test it directly. Re-run the same nested sampler with $\dot P$ removed from the likelihood (so the only data terms are $kT$ and $L_X$), keep all priors and forward models identical, and compare the marginals.

| Parameter | with $\dot P$ | spectral-only |
|:---|:---|:---|
| $M_{\rm WD}$ ($M_\odot$) | $0.87^{+0.13}_{-0.08}$ | $0.87^{+0.14}_{-0.08}$ |
| $\log_{10}\dot M$ | $14.72^{+0.03}_{-0.04}$ | $14.72^{+0.03}_{-0.04}$ |
| $\log_{10}\mu$ | $30.09\pm 0.34$ | $30.10\pm 0.34$ |
| $\xi$ | $0.57^{+0.27}_{-0.19}$ | $0.54^{+0.28}_{-0.18}$ |
| $\log_{10}(B/\text{G})$ | $3.76\pm 0.30$ | $3.79^{+0.31}_{-0.32}$ |

Every parameter shifts by less than $0.1\sigma$. **The hypothesis is empirically disconfirmed.** For CTCV J2056, the new $\dot P$ does not add parameter leverage on top of the spectral observables.

This is the section's headline finding, and the intent of the rest of the section (especially §9a) is to turn it from a passive observation into a constructive one. Three things matter for how it is interpreted.

**It is not a sampler artefact.** The corner overlay of with-$\dot P$ vs spectral-only posteriors is essentially identical, the evidence values agree within the $\sigma_{\ln Z}\sim 0.05$ floor, and the same null result reproduces under different random seeds.

**It is not a sign that the timing measurement is wasted.** The $\dot P$ likelihood term is still doing something: it enforces $|\dot P|\le \sigma_{\dot P}$, which is a non-trivial falsifiability constraint. Had the spectral fit predicted $|\dot P|\sim 10^{-13}$ (well above $\sigma_{\dot P}$), the test would have rejected the spectral picture. The non-detection rules that out. So $\dot P$ here functions as a consistency check on the equilibrium hypothesis, not as a parameter constraint.

**It admits a clean mechanistic explanation,** which is what §9a delivers via the prior-predictive check.

## 9. Validation

Posterior-predictive checks are a natural sanity test for any inference, but for this one they are *structural*: a 4-parameter model has more freedom than 3 observables, so pushing the posterior back through the same forward models must reproduce the data by construction (Gelman 2013, Example 1). To produce a non-trivial validation we need checks whose strength is not dependent on the same data going in twice. Notebook `09` runs three such checks.

### 9a. Prior predictive — and the mechanistic explanation for the leverage finding

The prior predictive distribution is what you get when you draw samples from the priors only (no likelihood), push them through the forward models, and look at the resulting distributions of $L_X$, $kT$, and $\dot P$. We draw $N=20{,}000$ prior samples, drop the $\sim 91\%$ rejected by the propeller / buried-field cut (the same admissibility cut the likelihood applies), and bin the survivors.

**$L_X$ and $kT$.** The prior predictive for $L_X$ spans roughly $10^{27}$-$10^{32}$ erg/s — six decades — with the observed $L_X\sim 1.8\times 10^{31}$ erg/s sitting somewhere inside the bulk. The prior predictive for $kT$ is broad (most of $0$-$80$ keV), with the observed 16.3 keV again well inside the bulk. The interpretation is unambiguous: the priors are *not* concentrated on the data, so the posterior tightness on $(M, \dot M)$ is being driven by the likelihood, not the prior. This is the direct test that the inference is doing genuine work, not just reflecting prior-dominated structure.

**$|\dot P|$ — the load-bearing piece.** The prior predictive for $|\dot P|$ tells a different story. A large fraction of prior samples have $|\dot P|$ already inside $\pm \sigma_{\dot P}$. Most of the prior-predictive distribution sits within an order of magnitude of the measurement floor.

**This is the equation-level explanation for the leverage finding.** $\dot P_{\rm pred}\propto P^2/I$, and $P=29.6$ s makes the natural scale at which $|\dot P|$ varies extremely small for any reasonable $(M, \dot M, \mu, \xi)$ in the accreting-IP regime; near-equilibrium torque cancellation around the equilibrium fastness $x_*<1$ then suppresses it further. Adding the $\dot P\approx 0$ likelihood term to a 30-second-spinner accreting-IP forward model has very little spectrally-allowed parameter space to down-weight, because the *forward model itself* doesn't let $|\dot P|$ get large enough to test against $\sigma_{\dot P}\sim 10^{-15}$ across most of the prior. The leverage failure is not a feature of the data or of the spectral observables; it is a feature of the $\dot P$ forward model evaluated at $P\sim 30$ s.

**This generalises.** The prior-predictive diagnostic now becomes a *predictor* of when pulsar-timing $\dot P$ should add parameter leverage in similar inferences:

- *Yes* — when the prior-predictive $|\dot P|$ distribution extends well beyond the achievable $\sigma_{\dot P}$ and the observed value lies in its tail. Examples: propeller spin-down systems with $|\dot P|\sim 10^{-13}$-$10^{-11}$ (e.g. AE Aqr), strong spin-up sources where the predicted $|\dot P|$ is dominated by accretion torque without near-cancellation.
- *No* — when the prior-predictive $|\dot P|$ is already concentrated near the measurement floor. This is CTCV J2056, and it will be true of any equilibrium-accreting IP with $P$ tens of seconds or longer. For these systems, $\dot P$ is best framed as a falsifiability check.

This is the methodological lesson the paper carries away from the negative result: the relationship between timing precision and parameter inference depends on the dynamical regime, and the prior-predictive distribution diagnoses the regime in advance.

### 9b. Independent physics consistency vs. Salcedo et al. 2024

A second validation is to compare our marginals on physical quantities to constraints derived without our likelihood. Salcedo et al. (2024) report, from independent X-ray spectral fits:

| Parameter | Our IP posterior | Salcedo+2024 |
|:---|:---|:---|
| $M_{\rm WD}$ ($M_\odot$) | $0.87^{+0.13}_{-0.08}$ | $0.7$-$1.0$ (Tab 6-7) |
| $\log_{10}\dot M$ (g/s) | $14.72^{+0.03}_{-0.04}$ | $14.36$-$14.95$ (§5.2) |
| $\log_{10}(B/\text{G})$ | $3.76\pm 0.30$ | $5.0$-$6.0$ (§6.3) |

$M_{\rm WD}$ and $\dot M$ agree — different methods, different data weightings, same answer. That is the consistency check passing: the inference reproduces the parts of Salcedo's analysis that are well-constrained.

The $\log_{10} B$ result *disagrees* by $\sim 2$ dex, which is not consistent with measurement noise. The disagreement is real and demands an explanation; that is what 9c provides.

### 9c. Vermette+2023 procedure cross-validation — locating the source of the $B$ disagreement

Salcedo §6.3 quotes $B\sim 0.1$-$1$ MG citing Vermette et al. (2023) and Norton et al. (2004). Vermette+2023 (Table 4, Case A) report $B=7$ MG for the IP J1745 with inputs $M=0.94\,M_\odot$, $r_m/R=20$, $\dot M=2\times 10^{16}$ g/s, $\xi=0.5$ (canonical disk-Alfvén). We can directly test our forward model against their procedure.

**Step 1 — does our forward model reproduce Vermette's J1745 result?** Inverting the Ghosh-Lamb $r_m=\xi(\mu^4/(2GM\dot M^2))^{1/7}$ for $\mu$ at the standard disk-accretion value $\xi=0.5$ and applying it to the J1745 inputs gives $\mu = (2GM)^{1/4}\dot M^{1/2} r_m^{7/4}/\xi^{7/4}$ and $B=\mu/R^3 = 4.1$ MG. Vermette's published value is 7 MG. Ours is a factor 1.7 smaller, which corresponds to an effective $\xi_{\rm eff}\approx 0.37$ — well inside our log-$\mathcal U(0.3, 1.0)$ prior range. The functional form is identical ($B\propto M\dot M^{1/2} r_m^{7/4}\xi^{-7/4}$); the small mismatch is an $\mathcal O(1)$ prefactor choice, exactly the systematic that our $\xi$ marginalisation already absorbs and that Norton et al. (2004) themselves explicitly characterise as order-of-magnitude only. **The forward model is the same standard-Alfvén procedure used in the cited literature.**

**Step 2 — apply the validated procedure to CTCV J2056.** Take Salcedo's MCVSPEC posterior values for J2056 (Table 7, Cases 2-3, with best-fit $\chi^2_\nu = 1.13$): $M=0.77$-$0.84\,M_\odot$, $r_m/R=1.8$-$2.0$ (under their $r_m=r_{\rm co}$ spin-equilibrium assumption), and back out $\dot M$ from $L_X = GM\dot M(1/R-1/r_m)$ at the observed low-state $L_X$. This yields $\dot M\sim 2$-$3\times 10^{14}$ g/s, fully consistent with Salcedo's quoted $(2.3$-$9.0)\times 10^{14}$ g/s low-state range. Plugging these inputs into the same Ghosh-Lamb procedure across our $\xi\in [0.3, 1.0]$ prior range gives $B\sim 5$-$30$ kG, fully consistent with our posterior median of $\sim 5.7$ kG.

**Why is J1745's $B$ in MG and J2056's $B$ in kG?** The difference is purely physical. $B\propto (R_m/R)^{7/4}\dot M^{1/2}$ at fixed $M$, $\xi$, $R$, and the relevant ratios are

$$\frac{B_{\rm J2056}}{B_{\rm J1745}} \sim \left(\frac{2}{20}\right)^{7/4}\!\left(\frac{2\times 10^{14}}{2\times 10^{16}}\right)^{1/2} \approx 0.018 \times 0.10 \approx 2\times 10^{-3},$$

i.e. $\sim 500\times$ smaller. CTCV J2056 should have a $B$ a few hundred times smaller than J1745. That is what we find.

**Conclusion.** Salcedo's quoted $B\sim 0.1$-$1$ MG for CTCV J2056 is *not* derivable from a careful application of the cited Vermette+2023 / Norton+2004 methodology to CTCV J2056's actual parameters. It appears to be a loose population-level comparison or extrapolation that did not rescale for J2056's much smaller magnetospheric radius and lower accretion rate. **Our $B$ posterior is the rigorous answer**, and it propagates the $\xi$ prefactor systematic that fixed-$\xi$ MCVSPEC fits do not. The $B$ "disagreement" with Salcedo is therefore not a contradiction at the methodology level; it is a quantification of a systematic that the previous analysis did not propagate.

## 10. Implications and conclusions

Pulling the threads together:

1. **Physical picture.** CTCV J2056 is a low-magnetic-field ($B\sim 5.7$ kG, factor-2 1σ envelope), evolved disk-fed intermediate polar in or near torque equilibrium. Its accretion rate is $\dot M\sim 5\times 10^{14}$ g/s; its WD mass is $\sim 0.87\,M_\odot$; its magnetospheric radius is $\sim 0.7\,r_{\rm co}$, sitting in the interior of the accreting regime at the equilibrium fastness predicted by the Bhattacharyya & Chakrabarty 2017 torque law. The spin-evolution timescale is $\sim 5$ Gyr, exceeding the Hubble time.

2. **Robustness.** The qualitative *near-equilibrium* statement is robust across the standard family of disk-torque prescriptions (GL79, Wang95, Rappaport+04, BC17). The specific numerical value $r_m/r_{\rm co}\sim 0.7$ and the $B\sim 5.7$ kG number are prescription-dependent at the $\mathcal O(10\text{-}30\%)$ level; the $\xi$ marginalisation absorbs an additional factor-3 prefactor systematic on $B$.

3. **The new $\dot P$ does not tighten parameter posteriors here.** The leverage test shows posteriors agree to $0.1\sigma$ with and without the $\dot P$ likelihood term. The prior-predictive check (§9a) explains why: $\dot P\propto P^2$ is naturally tiny at $P=30$ s, so the equation-level dynamic range of $|\dot P|$ across the spectrally-allowed parameter space is already below $\sigma_{\dot P}$.

4. **The $\dot P$ measurement is not wasted.** It still functions as an independent falsifiability check on the spectral picture — had spectral predictions implied $|\dot P|\gg \sigma_{\dot P}$, the test would have rejected the model. It is the right test to ask of an equilibrium-accretor candidate.

5. **The methodological generalisation — answer to the paper's "so what?" question.** Precision pulsar-timing $\dot P$ adds parameter leverage in CVs *away* from spin equilibrium. Concretely:
   - *Strong leverage:* propeller spin-down systems ($|\dot P|\sim 10^{-13}$-$10^{-11}$), strong spin-up systems where the predicted $|\dot P|$ is far above the achievable measurement floor. Examples include AE Aqr-class propellers and accretion-torque-dominated IPs.
   - *No leverage:* equilibrium accretors like CTCV J2056. For these, spectral observables dominate and timing is best framed as a consistency check.

   The prior-predictive distribution of $|\dot P|$ is the diagnostic — it can be computed in advance from the spectral-data prior alone, before committing to a tight $\dot P$ campaign.

6. **The $B$ disagreement with Salcedo is methodological, not contradictory.** Their MG-range quote is not reproducible from the cited Vermette/Norton procedure when actually applied to J2056's parameters. Our posterior is the rigorous answer because it propagates the $\xi$ prefactor systematic and the joint $(M, \dot M)$ covariance through a single likelihood.

The section therefore turns the negative leverage result into a constructive contribution: a calibrated statement about when this paper's timing methods translate into parameter improvements vs. when they should be sold as falsifiability tests, plus a defensible $B$ posterior and spin-equilibrium characterisation for CTCV J2056 specifically.

## 11. Caveats

Tracked here so they can be recited without re-deriving.

- **$kT_{\rm obs}$ from MKCFLOW.** Treated as approximately model-independent (high-energy continuum cutoff). A fully rigorous analysis would refit Salcedo's spectra with a unified IP model. Out of scope.
- **$kT$-$L_X$ covariance from shared MKCFLOW systematics is not propagated.** Likely tightens the posterior slightly more than is warranted in the $kT$-$L_X$ plane.
- **Single-temperature $f_{\rm band}$.** The post-shock column is a multi-temperature cooling flow; the single-$T$ approximation introduces a $\sim 20$-$30\%$ systematic on $L_{X,\rm pred}$ that the inference absorbs into the $\dot M$ posterior.
- **Propeller exclusion is a hard prior.** Physically motivated by the column-accretion evidence, but a hard cut.
- **$\xi$ prior range $[0.3, 1.0]$.** The dominant $B$ systematic is marginalised, but the prior range is itself a choice made on physical grounds (Ghosh & Lamb 1979 disk-Alfvén range). Effective values outside this range — e.g. the $\xi_{\rm eff}\approx 0.37$ implied by reproducing Vermette's J1745 result — sit at the lower boundary of our prior.
- **Torque-prescription dependence.** $r_m/r_{\rm co}\sim 0.7$ and the implied $B$ carry an additional $\mathcal O(10\text{-}30\%)$ systematic from the BC17-vs-Wang95-vs-GL79 prescription choice; not propagated. The qualitative near-equilibrium statement is robust.
